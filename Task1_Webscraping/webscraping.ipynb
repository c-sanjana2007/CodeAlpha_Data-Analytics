Importing the Libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np

print("All libraries imported successfully!")
```

Connecting to the Websites

```python
url = "https://books.toscrape.com/"
response = requests.get(url, timeout=15)
print("Status Code:", response.status_code)
```

Succesfully website connected

```python

# Website URL
url = "https://books.toscrape.com/"

# Get webpage content
response = requests.get(url)

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

# Store scraped data
books = []

# Extract book details
for book in soup.select("article.product_pod"):
    title = book.h3.a["title"]
    price = book.select_one(".price_color").get_text(strip=True)
    availability = book.select_one(".availability").get_text(strip=True)
    rating = book.select_one("p.star-rating")["class"][1]

    books.append({
        "Title": title,
        "Price": price,
        "Availability": availability,
        "Rating": rating
    })

print("Total books scraped:", len(books))
```

Converting scraped data into a DataFrame

```python
df = pd.DataFrame(books)
df.head()
```

Checking the dataset

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

Saving the dataset

```python
df.to_csv("scraped_books.csv", index=False)

print("Dataset saved successfully!")
```

Analyzing the scraped dataset

```python
# Total number of books
print("Total Books:", len(df))

# Book ratings count
print("\nRating Distribution:")
print(df["Rating"].value_counts())

# Availability count
print("\nBook Availability:")
print(df["Availability"].value_counts())
```

Checking the Price in data

```python

# Clean price data
df["Price"] = (
    df["Price"]
    .str.replace("Â", "", regex=False)
    .str.replace("£", "", regex=False)
    .str.strip()
    .astype(float)
)

# Calculate price statistics
print("Average Price: £", round(df["Price"].mean(), 2))
print("Minimum Price: £", df["Price"].min())
print("Maximum Price: £", df["Price"].max())
```

Saved the corrected Dataset

```python
df.to_csv("scraped_books.csv", index=False)

print("Updated dataset saved successfully!")
```
